ATIVIDADE I


```python
def grava_arquivo(num_acertar, tentativas):
  with open('/content/sample_data/tentativas.txt', 'a') as file:
    file.writelines([f'Número sorteado {num_acertar}', '\n'])
    file.writelines([f'Tentativas até o acerto {tentativas}', '\n'])
    file.close
```

```python
import random

rejogar = None

while rejogar != 0:

  num_acertar = random.randint(0, 5)
  tentativa = None
  tentativas = 0

  while tentativa != num_acertar:
    tentativas += 1
    tentativa = int(input("Tente acertar o número sorteado: "))

  grava_arquivo(num_acertar, tentativas)

  rejogar = int(input("Deseja continuar jogando? Digite 0 para não "))
```

ATIVIDADE II

```python
import csv

lista_compras = []

i = 1

while True:
  print(f'Produto {i}')
  qtd = int(input("Informe a quantidade de itens a ser adicionados a lista. Caso não queira adicionar mais nenhum digite '0'."))
  if qtd == 0:
    break
  valor = float(input("Informe o valor unitário do produto. "))

  lista_compras.append({
      "Produto": i,
      "Valor": valor,
      "Quantidade": qtd
  })

valor_ttl = 0

for item in lista_compras:
  valor_ttl += item['Valor'] * item['Quantidade']

print(f'Valor total da lista de compras {valor_ttl}')

with open('/content/sample_data/lista_compras.csv', 'w') as file:
  writer = csv.DictWriter(file, fieldnames=['Produto', 'Valor', 'Quantidade'])
  writer.writeheader()
  writer.writerows(lista_compras)
```

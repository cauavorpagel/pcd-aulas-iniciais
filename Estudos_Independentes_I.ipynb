Desafio 1 – Análise de notas de uma turma
Desenvolva um programa que permita registrar as notas de uma turma de estudantes.
O programa deverá:
* solicitar a quantidade de estudantes;
* armazenar as notas em uma lista;
calcular:
* maior nota;
* menor nota;
* média da turma;
* quantidade de estudantes aprovados (nota ≥ 6,0);
* quantidade de estudantes reprovados.
Os cálculos deverão ser implementados utilizando funções.

```python
def calc_maior(notas):
  return max(notas)

def calc_menor(notas):
  return min(notas)

def calc_media(notas):
  return sum(notas) / len(notas)

def verifica_aprovacao(notas):
  aprovados = 0
  reprovados = 0
  for nota in notas:
        if nota >= 6.0:
            aprovados += 1
        else:
            reprovados += 1
  return aprovados, reprovados

lista_notas = []

qtd_alunos = int(input("informe a quantidade de alunos na turma:"))

for i in range(qtd_alunos):
  nota_aluno = float(input("informe a nota do aluno: "))
  lista_notas.append(nota_aluno)

maior_nota = calc_maior(lista_notas)
menor_nota = calc_menor(lista_notas)
media_notas = calc_media(lista_notas)
aprovados, reprovados = verifica_aprovacao(lista_notas)

print(f"Notas: {lista_notas}")
print(f"Maior nota: {maior_nota:.1f}")
print(f"Menor nota: {menor_nota:.1f}")
print(f"Média da turma: {media_notas:.2f}")
print(f"Alunos aprovados: {aprovados}")
print(f"Alunos reprovados: {reprovados}")
```

Desafio 2 – Cadastro e consulta de produtos
Desenvolva um programa que permita cadastrar produtos utilizando um dicionário.
Cada produto deverá possuir:
* código;
* nome;
* preço;
* quantidade em estoque.
Ao final, o programa deverá permitir:
* consultar um produto pelo código;
* calcular o valor total do estoque;
* identificar o produto de maior valor unitário.
Organize a solução utilizando funções.

```python
def registrar_produto():
    produtos = {}
    qtd = int(input("Quantos produtos você deseja cadastrar? "))

    for i in range(qtd):
        cod = input(f'Informe o código do produto {i + 1}: ')
        nome = input("Informe o nome do produto: ")
        preco = float(input("Informe o preço do produto: "))
        quantidade = int(input("Informe a quantidade a ser armazenada: "))

        produtos[cod] = {
            "nome": nome,
            "preco": preco,
            "quantidade": quantidade
        }

    return produtos

def consulta_produto(lista_produtos):
    cod = input("Informe o código do produto a ser pesquisado: ")
    if cod in lista_produtos:
        item = lista_produtos[cod]
        print(f"Nome: {item['nome']}")
        print(f"Preço: R$ {item['preco']:.2f}")
        print(f"Estoque: {item['quantidade']} unidades")
    else:
        print("Produto não encontrado.")

def valor_total_estoque(lista_produtos):
    total = sum(item["preco"] * item["quantidade"] for item in lista_produtos.values())
    return total

def identificar_maior_valor_unitario(lista_produtos):
    codigo_maior = max(lista_produtos, key=lambda k: lista_produtos[k]["preco"])
    return codigo_maior, lista_produtos[codigo_maior]

estoque = registrar_produto()

if estoque:
    valor_total = valor_total_estoque(estoque)
    print(f"\nValor total do estoque: R$ {valor_total:.2f}")

    cod_maior, item_maior = identificar_maior_valor_unitario(estoque)
    print(f"Produto de maior valor unitário: {item_maior['nome']} (Código: {cod_maior}) - R$ {item_maior['preco']:.2f}\n")

    consulta_produto(estoque)
else:
    print("Nenhum produto cadastrado.")
```

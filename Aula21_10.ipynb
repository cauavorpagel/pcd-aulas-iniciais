```python
import pandas as pd
import numpy as np

nomes = ['SepalLength', 'SepalWidth', 'PetalLength', 'PetalWidth', 'Class']

df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data', names = nomes)

print(f'\n{df}')
```

```python
X = df[df.columns.difference(['Class'])].values
print(f'X: {X}')

Y = df['Class'].values
print(f'Y: {Y}')
```

```python
from sklearn.neighbors import KNeighborsClassifier

flor_iris_classificador = KNeighborsClassifier(n_neighbors=3)

flor_iris_classificador.fit(X,Y)
```

```python
from sklearn.model_selection import cross_val_score

scores_dt = cross_val_score(flor_iris_classificador, X, Y, scoring='accuracy', cv=5)

print(scores_dt.mean())
```

```python
for i in range(1, 11):
  flor_iris_classificador = KNeighborsClassifier(n_neighbors=i)
  flor_iris_classificador.fit(X,Y)
  scores_dt = cross_val_score(flor_iris_classificador, X, Y, scoring='accuracy', cv=5)
  print(f'Vizinhos: {i} Acuracia: {scores_dt.mean()}')
```

```python
novos_exemplos = [[5.2, 2.4, 7.0, 3.2], [4.2, 1.2, 5.8, 2.7], [1.6, 0.5, 5.0, 3.6]]
print(flor_iris_classificador.predict(novos_exemplos))
```

```python
sepal_length = input('Comprimento da sépala: ')
sepal_width = input('Largura da sépala: ')
petal_length = input('Comprimento da Pétala: ')
petal_width = input('Largura da Pétala: ')
teste = [[float(sepal_length), float(sepal_width), float(petal_length), float(petal_width)]]
print(f'Amostra Classificada como: {flor_iris_classificador.predict(teste)}')
```

```python
colunas = ['Acidez Fixa','Acidez Volátil','Acidez Citríca','Açúcar Residual','Cloreto','Livre de Dióxido de Enxofre','Dióxido de Enxofre Total','Densidade','pH','Sulfato','Alcool','Qualidade']
df_vinhos = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv', names = colunas, sep=';', skiprows=1)
df_vinhos.head()
```

```python
condicoes = [df_vinhos['Qualidade'] >= 7.0, df_vinhos['Qualidade'] >= 5.0, df_vinhos['Qualidade'] < 5.0]
notas = ['Alta', 'Média', 'Baixa']

df_vinhos['Classificação'] = np.select(condicoes, notas, default="Desconhecido")

df_vinhos.head()
```

```python
X_vinho = df_vinhos[df_vinhos.columns.difference(['Classificação', 'Qualidade'])].values
print(f'X: {X_vinho}')

Y_vinho = df_vinhos['Classificação'].values
print(f'Y: {Y_vinho}')
```

```python
vinho_classificador = KNeighborsClassifier(n_neighbors=67)

vinho_classificador.fit(X_vinho,Y_vinho)
```

```python
score_dt = cross_val_score(vinho_classificador, X_vinho ,Y_vinho, scoring='accuracy', cv=5)

print(score_dt.mean())
```

```python
for i in range(1, 11):
  vinho_classificador = KNeighborsClassifier(n_neighbors=i)
  vinho_classificador.fit(X_vinho,Y_vinho)
  score_dt = cross_val_score(vinho_classificador, X_vinho, Y_vinho, scoring='accuracy', cv=5)
  print(f'Vizinhos: {i} Acuracia: {score_dt.mean()}')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dados = {
    "Mes": ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun"],
    "Notebook": [20, 25, 22, 30, 35, 40],
    "Mouse": [50, 55, 48, 60, 65, 70],
    "Teclado": [30, 28, 35, 40, 38, 45]
}

df = pd.DataFrame(dados)
print(df)
```

```python
#1. Exploração: Exiba as primeiras linhas, as dimensões do DataFrame e as estatísticas descritivas.
print(df.head())
print("\n")
print(f"Dimensões: {df.shape}")
print(f"Linhas: {df.shape[0]}")
print(f"Colunas: {df.shape[1]}")
print("\n")
df.describe()
```

```python
#2. Indicadores: Calcule com NumPy a média, o maior valor e o menor valor de vendas de cada produto.

print("Notebook")
med_notebook = np.mean(df.Notebook)
maior_notebook = np.max(df.Notebook)
menor_notebook = np.min(df.Notebook)

print(f"Média de preço: R${med_notebook:.2f}")
print(f"Maior preço: R${maior_notebook:.2f}")
print(f"Menor preço: R${menor_notebook:.2f}")

print("\n")

print("Mouse")
med_mouse = np.mean(df.Mouse)
maior_mouse = np.max(df.Mouse)
menor_mouse = np.min(df.Mouse)

print(f"Média de preço: R${med_mouse:.2f}")
print(f"Maior preço: R${maior_mouse:.2f}")
print(f"Menor preço: R${menor_mouse:.2f}")

print("\n")

print("Teclado")
med_teclado = np.mean(df.Teclado)
maior_teclado = np.max(df.Teclado)
menor_teclado = np.min(df.Teclado)

print(f"Média de preço: R${med_teclado:.2f}")
print(f"Maior preço: R${maior_teclado:.2f}")
print(f"Menor preço: R${menor_teclado:.2f}")
```

```python
#3. Evolução temporal: Crie um gráfico de linhas comparando as vendas dos três produtos ao longo dos meses.
x = df.Mes
yN = df.Notebook
yM = df.Mouse
yT = df.Teclado

plt.grid(color="black")
plt.ylabel("Quantidade de produtos")
plt.xlabel("Mês")
plt.plot(x, yN)
plt.plot(x, yM)
plt.plot(x, yT)
plt.show()
```

```python
#4. Comparação: Crie um gráfico de barras com o total vendido por produto no período.
produtos = ["Notebook", "Mouse", "Teclado"]
totais = []
for produto in produtos:
    totais.append(int(np.sum(df[produto])))

print("Totais:", totais)

plt.figure(figsize=(6, 4))
plt.bar(produtos, totais)
plt.title("Total vendido por produto")
plt.ylabel("Unidades vendidas")
plt.show()
```

```python
#5. Distribuição: Construa um histograma com os valores mensais de vendas de um produto escolhido.
plt.figure(figsize=(6, 4))
plt.hist(df["Teclado"], bins=4, edgecolor="black")
plt.title("Distribuição das vendas mensais de teclados")
plt.xlabel("Unidades vendidas no mês")
plt.ylabel("Quantidade de meses")
plt.show()
```

```python
#6. Interpretação: Identifique o produto com maior volume total, o mês de maior venda de cada produto e uma conclusão baseada nos gráficos.
posicao = np.argmax(totais)   # argmax devolve a posição do maior valor
print("Produto com maior volume total:", produtos[posicao], "com", totais[posicao], "unidades")
print()

# mês de maior venda de cada produto
for produto in produtos:
    linha = df[produto].idxmax()     # índice da linha com o maior valor
    mes = df["Mes"][linha]
    print("Maior venda de", produto, "foi em", mes, "com", df[produto][linha], "unidades")

teclado = np.array(df["Teclado"])
meses = list(df["Mes"])

variacoes = []
meses_variacao = []

# começo no 1 porque janeiro não tem mês anterior para comparar
for i in range(1, len(teclado)):
    variacao = (teclado[i] - teclado[i - 1]) / teclado[i - 1] * 100
    variacoes.append(round(variacao, 1))
    meses_variacao.append(meses[i])

for i in range(len(variacoes)):
    print(meses_variacao[i], ":", variacoes[i], "%")

cores = []
for v in variacoes:
    if v >= 0:
        cores.append("green")
    else:
        cores.append("red")

plt.figure(figsize=(7, 4))
plt.bar(meses_variacao, variacoes, color=cores)
plt.axhline(0, color="black")   # linha do zero, separa crescimento de queda
plt.title("Teclado: variação percentual em relação ao mês anterior")
plt.xlabel("Mês")
plt.ylabel("Variação (%)")
plt.show()
```

```python
#Agora, escolha um dos produtos e crie uma nova visualização que responda a uma pergunta diferente daquelas já exploradas.
#Pode ser uma comparação entre produtos, uma análise de variação percentual ou a identificação de meses com resultados abaixo da média. Explique por que escolheu o tipo de gráfico e qual informação ele permite observar.

# 7. Vendas de notebooks abaixo da média

notebooks = df["Notebook"]
media = np.mean(notebooks)

cores = ["red" if venda < media else "green" for venda in notebooks]

plt.figure(figsize=(8, 5))
plt.bar(df["Mes"], notebooks, color=cores)
plt.axhline(media, color="blue", linestyle="--",
            label=f"Média mensal: {media:.2f}")

plt.title("Vendas de Notebooks em Relação à Média")
plt.xlabel("Meses")
plt.ylabel("Quantidade Vendida")
plt.legend()
plt.show()
```

```python
import google_colab_selenium as gs
from selenium.webdriver.common.by import By
from time import sleep
```

```python
navegador = gs.Chrome()
```

```python
cidade = input("Digite a cidade que você deseja ver o clima-tempo.")
estado = input("Digite a sigla do estado da cidade.")

```

```python
url = (
    f"https://hgbrasil.com/weather/places/brazil/{estado}/{cidade}"
)
```

```python
navegador.get(url)
sleep(2)
print("Página:", navegador.title)
```

```python
import re
def remove_non_numeric_from_str(value: str):
  return re.sub("[^0-9]", "", value)
```

```python
temperaturas = remove_non_numeric_from_str(
    navegador.find_element(
    By.XPATH,
    '//*[@id="main"]/article/div/div/section[1]/div[4]/div[2]/span[1]'
).text)

print("A temperatura máxima na cidade de", cidade, "é" ,temperaturas)


```

```python
minima = remove_non_numeric_from_str(
    navegador.find_element(
    By.XPATH,
    '//*[@id="main"]/article/div/div/section[1]/div[4]/div[2]/span[2]'
).text)

print("A temperatura minima na cidade de", cidade, "é" ,temperaturas)
```

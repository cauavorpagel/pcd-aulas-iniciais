```python
nomeAluno = input("Digite seu nome: ")
nota1 = float(input("Digite a sua nota na primeira prova: "))
nota2 = float(input("Digite a sua nota na segunda prova: "))
nota3 = float(input("Digite a sua nota na terceira prova: "))

media = (nota1 + nota2 + nota3) / 3

print("Nome do aluno: ", nomeAluno)
print("Nota da primeira prova: ", nota1)
print("Nota da segunda prova: ", nota2)
print("Nota da terceira prova: ", nota3)
print("Média total do aluno", f"{media: .1f}")
```

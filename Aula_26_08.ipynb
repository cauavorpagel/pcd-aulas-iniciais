JSON

```python
import json

pessoa = '{"nome": "Pedro", "linguagens": ["Espanhol", "Português"]}'

pessoa_dic = json.loads(pessoa)

print(pessoa_dic)

print(pessoa_dic['linguagens'])
```

```python
import json

with open('sample_data/anscombe.json', 'r') as f:
  dados = json.load(f)

print(dados)

print(json.dumps(dados, indent = 4, sort_keys=True))
```

```python
pessoa_add = {"casado": True,"idade": 32}

pessoa_dic.update(pessoa_add)

with open('sample_data/pessoa.txt', 'w') as json_file:
  json.dump(pessoa_dic, json_file, indent=4)
```

```python
import json
import requests
```

```python
n = input("Digite o nome do personagem: ")
lista_poderes = []

while True:
  p = input("Digite o poder que você quer adicionar: ")
  lista_poderes.append(p)
  f = input("Deseja adicionar mais um? s ou n ")
  if f == "n":
    break

personagem_add = {"Nome": n, "Poderes": lista_poderes}

with open('sample_data/personagem.txt', 'w') as json_file:
  json.dump(personagem_add, json_file, indent=4)

print(personagem_add)

```

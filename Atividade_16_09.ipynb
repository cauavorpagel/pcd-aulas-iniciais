```python
import pandas as pd

# Variável que recebe os dados climatológicas via GitHub
url = "https://raw.githubusercontent.com/turicas/rows/master/examples/brazil-climate/data/climate_data.csv"

# Lê o CSV, possui uma URL secundária para caso a primeira oscile na leitura
try:
    df_raw = pd.read_csv(url)
except Exception:
    url = "https://raw.githubusercontent.com/datasets/climate-change/master/data/global-temp.csv"

# Cria a estrutura com os dados climáticos
dados_santa_rosa = {
    'Cidade': ['Santa Rosa'] * 12,
    'Mes': ['Janeiro', 'Fevereiro', 'Março', 'Abril', 'Maio', 'Junho',
            'Julho', 'Agosto', 'Setembro', 'Outubro', 'Novembro', 'Dezembro'],
    'Num_Mes': list(range(1, 13)),
    'Estacao': ['Verão', 'Verão', 'Outono', 'Outono', 'Outono', 'Inverno',
                'Inverno', 'Inverno', 'Primavera', 'Primavera', 'Primavera', 'Verão'],
    'Temperatura_Media': [25.1, 24.3, 22.8, 19.5, 16.2, 14.8, 14.1, 15.9, 17.5, 20.2, 22.4, 24.5],
    'Temperatura_Minima': [19.2, 18.9, 17.1, 14.3, 11.5, 9.8, 8.9, 10.2, 12.1, 14.8, 16.5, 18.1],
    'Deficit_Hidrico_mm': [18.0, 12.0, 5.0, 0.0, 0.0, 0.0, 0.0, 22.0, 0.0, 0.0, 0.0, 15.0]
}

df_cidade = pd.DataFrame(dados_santa_rosa)

print("=== 1. TODAS AS INFORMAÇÕES DE SANTA ROSA - RS ===")
print(df_cidade[['Mes', 'Estacao', 'Temperatura_Media', 'Temperatura_Minima', 'Deficit_Hidrico_mm']])

idx_min_temp = df_cidade['Temperatura_Minima'].idxmin()
mes_min_temp = df_cidade.loc[idx_min_temp, 'Mes']
valor_min_temp = df_cidade.loc[idx_min_temp, 'Temperatura_Minima']

print("\n=== 2. MÊS DE TEMPERATURA MÍNIMA ===")
print(f"Mês: {mes_min_temp} (Mínima de {valor_min_temp}°C)")

idx_max_deficit = df_cidade['Deficit_Hidrico_mm'].idxmax()
mes_max_deficit = df_cidade.loc[idx_max_deficit, 'Mes']
valor_max_deficit = df_cidade.loc[idx_max_deficit, 'Deficit_Hidrico_mm']

print("\n=== 3. MÊS DE MAIOR DÉFICIT HÍDRICO ===")
print(f"Mês: {mes_max_deficit} (Déficit de {valor_max_deficit} mm)")

media_temp_estacoes = df_cidade.groupby('Estacao')['Temperatura_Media'].mean()

print("\n=== 4. MÉDIA DE TEMPERATURA POR ESTAÇÃO DO ANO ===")
print(media_temp_estacoes)
```

IF, Else e Elif

```python
nota = float(input("Digite o valor da nota:"))
if nota < 60:
  print("Sua nota {} foi insuficiente".format(nota))
```

```python
nota = float(input("Digite o valor da nota: "))
if nota < 60:
  print("Sua nota {} foi insuficiente.".format(nota))
  print("Nota ruim!")
else:
  print("Sua nota {} foi suficiente.".format(nota))
  print("Nota boa!")
```

```python
nota = float(input("Digite o valor da nota: "))
if nota < 60:
  print("Sua nota {} foi insuficiente.".format(nota))
elif nota >= 60 and nota < 75:
  print("Sua nota {} foi C.".format(nota))
elif nota >= 75 and nota < 90:
  print("Sua nota {} foi B.".format(nota))
else:
  print("Sua nota {} foi A.".format(nota))
```

While

```python
senha = "112233"
digitado = " "
while(digitado != senha):
  digitado = input("Digite a senha: ")
  if digitado == senha:
    print("Acesso Liberado!")
  else:
    print("Senha incorreta. Tente novamente!")
```

For

```python
for i in range(0, 11, 1):
  print(i)
```

```python
lista_impares = []
for i in range(20, 10, -1):
  if i % 2 != 0:
    lista_impares.append(i)
  print(lista_impares)
```

For em listas, Tuplas e Dicionários

```python
notas_provas = [9, 6.5, 3.4, 8,5]
soma = 0
for nota in notas_provas:
  soma = soma + nota

media = soma / 4
print("Média = ", media)
```

Functions


```python
def harry():
  return("VAI ESTABILIZAR! ESTÁ SOB CONTROLE!")

print("DESLIGA ISSO OTTO! DESLIGAAA!!!\n" + harry())
```

```python
def maior(a, b):
  if a > b:
    print("Maior: {}".format(a))
  else:
    print("Maior: {}".format(b))

maior(120, 35)
maior(35, 120)
```

```python
def calcula_percento(valor, percentual = 100):
  quanto = valor * percentual / 100
  return quanto

print("20% de 300 são {:10.2f}".format(calcula_percento(300, 20)))
print("100% de 50 são {:10.2f}".format(calcula_percento(50)))
```

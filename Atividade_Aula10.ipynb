```python
import numpy as np

dados = np.array([
                   [120, 15000, 100, 125.00, 4],
                   [135, 16200, 108, 120.00, 5],
                   [98, 11700, 82, 119.39, 3],
                   [150, 21000, 125, 140.00, 7],
                   [175, 24500, 140, 140.00, 9],
                   [160, 22400, 132, 140.00, 6],
                   [110, 13200, 95, 120.00, 2]
])

dias = np.array([
"Segunda",
"Terça",
"Quarta",
"Quinta",
"Sexta",
"Sábado",
"Domingo"
])
```

Parte 1 — Conhecendo os dados

```python
linhas_colunas = dados.shape
total = dados.size
dimensoes = dados.ndim
tipo_dados = dados.dtype

print(f'Linhas: {linhas_colunas[0]}')
print(f'Colunas: {linhas_colunas[1]}')
print(f'Total de itens: {total}')
print(f'Dimensões: {dimensoes}')
print(f'Tipo dos dados: {tipo_dados}')
```

Parte 2 — Acessando os dados

```python
dados_tratados = np.column_stack((dados, dias))

print(dados_tratados)
```

```python
print(f'Dados da Sexta-Feira {dados_tratados[4]}')
print(f'Faturamento de todos os dias: {dados_tratados[:, 1]}')
total_reclamacoes = np.sum(dados[:, 4])
print(f'Total de reclamações: {total_reclamacoes}')
print(f'Dados de Quinta a Domingo: \n{dados_tratados[3:]}')
print(f'Colunas: pedidos, faturamento, clientes: \n{dados_tratados[:, 0:3]}')
```

Parte 3 — Operações numéricas

```python
total_pedidos = np.sum(dados[:, 0])
print(f'Total de pedidos realizados: {total_pedidos}')

total_faturamento = np.sum(dados[:, 1])
print(f'Total de faturamento: R${total_faturamento}')

total_clientes = np.sum(dados[:, 2])
print(f'Total de clientes: {total_clientes}')

media_pedidos = np.round((total_pedidos / 7), 2)
print(f'Média de pedidos por dia: {media_pedidos}')

media_faturamento = np.round((total_faturamento / 7), 2)
print(f'Média de faturamento: R${media_faturamento}')

media_reclamacoes = np.round((total_reclamacoes / 7), 2)
print(f'Media de reclamções por dia: {media_reclamacoes}')
```

Parte 4 — Análise por dia

```python
maior_num_pedidos = np.argmax(dados[:, 0])
menor_num_pedidos = np.argmin(dados[:, 0])

print(f'Dia com o maior número de pedidos: {dias[maior_num_pedidos]}')
print(f'Dia com o menor número de pedidos: {dias[menor_num_pedidos]}')

maior_faturamento = np.argmax(dados[:, 1])
menor_faturamento = np.argmin(dados[:, 1])

amplitude = dados[maior_faturamento, 1] - dados[menor_faturamento, 1]

print(f'Dia com o maior faturamento: {dias[maior_faturamento]}')
print(f'Dia com o menor faturamento: {dias[menor_faturamento]}')
print(f'A diferença entre esses dois dias foi de: R${amplitude}')
```

Parte 5 — Máscaras e filtros

```python
print(f'Dias com mais de RS12000,00 em faturamento: {dias[np.where(dados[:, 1] >= 12000)]}')
print(f'Dias com mais de 150 pedidos: {dias[np.where(dados[:, 0] >= 150)]}')
print(f'Dias com mais de RS12000,00 em faturamento: {dias[np.where(dados[:, 4] >= 4)]}')
print(f'Faturamento dos dias com mais de 150 pedidos: {dados_tratados[np.where(dados[:, 0] >= 150)][:, 1]}')
print(f'Dias com faturamento abaixo da média semanal: {dias[np.where(dados[:, 1] < media_faturamento)]}')
```

Parte 6 — Classificação dos dias

```python
classificacao = np.where(
    dados[:, 1] >= 20000,
    "Alto desempenho",
    "Desempenho normal"
)

index_dia_alto_desempenho = np.where(classificacao == 'Alto desempenho')
index_dia_desempenho_normal = np.where(classificacao == 'Desempenho normal')

print(f'Dias com alto desempenho: {len(dias[index_dia_alto_desempenho])} dias. {dias[index_dia_alto_desempenho]}')
print(f'Dias com alto desempenho: {len(dias[index_dia_desempenho_normal])} dias. {dias[index_dia_desempenho_normal]}')
```

Parte 7 — Estatística

```python
media = np.mean(dados[:, 1])
mediana = np.median(dados[:, 1])
desvio_padrao = np.std(dados[:, 1])
index_max = np.argmax(dados[:, 1])
max = dados[index_max, 1]
index_min = np.argmin(dados[:, 1])
min = dados[index_min, 1]


print("Média: ", media)
print("Mediana: ",mediana)
print("Desvio padrão: ", desvio_padrao)
print("Máximo: ", max)
print("Mínimo: ", min)
```

Parte 8 — Trabalhando com axis

```python
media_indicadores = np.mean(dados, axis=0)
max_indicadores = np.max(dados, axis=0)
min_indicadores = np.min(dados, axis=0)

variacao_relativa_indicadores = np.diff(dados, axis=0) / dados[:-1]
maior_oscilacao_por_coluna = np.max(np.abs(variacao_relativa_indicadores), axis=0)
indicador_vencedor = np.argmax(maior_oscilacao_por_coluna)

print(f"Média de pedidos: {media_indicadores[0]}")
print(f"Média de faturamento: {media_indicadores[1]}")
print(f"Média de clientes: {media_indicadores[2]}")
print(f"Média de valor médio dos pedidos: {media_indicadores[3]}")
print(f"Média de reclamações: {media_indicadores[4]}")
print("\n")
print(f"Maior de pedidos: {max_indicadores[0]}")
print(f"Maior de faturamento: {max_indicadores[1]}")
print(f"Maior de clientes: {max_indicadores[2]}")
print(f"Maior de valor médio dos pedidos: {max_indicadores[3]}")
print(f"Maior de reclamações: {max_indicadores[4]}")
print("\n")
print(f"Menor de pedidos: {min_indicadores[0]}")
print(f"Menor de faturamento: {min_indicadores[1]}")
print(f"Menor de clientes: {min_indicadores[2]}")
print(f"Menor de valor médio dos pedidos: {min_indicadores[3]}")
print(f"Menor de reclamações: {min_indicadores[4]}")
print("\n")
print(f"Maior variação relativa: {indicador_vencedor}")
```

Parte 9 — Análise integrada

```python
index_dia_melhor_faturamento = np.argmax(dados[:, 1])
index_dia_mais_pedidos = np.argmax(dados[:, 0])
index_dia_mais_reclamacoes = np.argmax(dados[:, 4])

print(f'Dia com maior faturamento: {dias[index_dia_melhor_faturamento]}')

if index_dia_melhor_faturamento == index_dia_mais_pedidos:
  print("O dia com mais faturamento foi o dia com mais pedidos")
else:
  print(f"O dia com mais pedidos foi: {dias[index_dia_mais_pedidos]}")

if index_dia_melhor_faturamento == index_dia_mais_reclamacoes:
  print("O dia com mais faturamento foi o dia com mais reclamações")
else:
  print(f"O dia com mais pedidos foi: {dias[index_dia_mais_reclamacoes]}")

dias_com_mais_150_pedidos = dias[np.where((dados[:, 0] >= 150) & (dados[:, 1] >= 20000))]

print(dias_com_mais_150_pedidos)
```

```python
print("===================RESUMO SEMANAL====================")
print("\n")
print(f"Total de pedidos: {total_pedidos}")
print(f"Faturamento total: {total_faturamento}")
print(f"Média diária de pedidos: {media_pedidos}")
print(f"Média diária de faturamento: {media_indicadores[1]}")
print(f"Melhor dia: {dias[index_dia_melhor_faturamento]}")
print(f"Pior dia: {dias[np.argmin(dados[:, 1])]}")
print(f"Maior número de reclamações: {np.max(dados[:, 4])}")
print(f"Dias de alto desempenho: {dias[index_dia_alto_desempenho]}")
print("\n")
print("=====================================================")






```

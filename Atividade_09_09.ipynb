```python
class Calculadora():
  def __init__(self):
    return None

  def somar(self, x, y):
    return x + y

  def subtracao(self, x, y):
    return x - y

  def multiplicacao(self, x, y):
    return x * y

  def divisao(self, x, y):
    return x / y

  def exponenciacao(self, x, y):
    return x ** y

  def p(self, b, n):
    return (b + (n/b) ) / 2

  def raiz(self, n):
    b = 1
    while True:
      b = self.p(b, n)
      if (b ** 2) - n < 0.0001:
        break

    return b
```

```python
calculadora = Calculadora()
operacao = int(input("Soma = 1, Subtração = 2, Multiplicação = 3, Divisão = 4, Exponenciação = 5, Raiz = 6 \n"))

resultado = 0
if operacao == 1:
  x = float(input("Digite um num: "))
  y = float(input("Digite um num: "))
  resultado = calculadora.somar(x, y)
elif operacao == 2:
  x = float(input("Digite um num: "))
  y = float(input("Digite um num: "))
  resultado = calculadora.subtracao(x, y)
elif operacao == 3:
  x = float(input("Digite um num: "))
  y = float(input("Digite um num: "))
  resultado = calculadora.multiplicacao(x, y)
elif operacao == 4:
  x = float(input("Digite um num: "))
  y = float(input("Digite um num: "))
  resultado = calculadora.divisao(x, y)
elif operacao == 5:
  x = float(input("Digite um num: "))
  y = float(input("Digite um num: "))
  resultado = calculadora.exponenciacao(x, y)
elif operacao == 6:
  x = float(input("Digite um num: "))
  resultado = calculadora.raiz(x)

print(resultado)
```
